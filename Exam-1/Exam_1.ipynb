#### Q2. Use the cell below to enter your code for this question.
Suppose that you have a string ab_string = ‘ababababab’. Write an expression to remove all the ‘b’s and create a string, a_string = ‘aaaaa’.

```python
ab_string = 'ababababab'
new_string = ''
b_string = ''

for i in ab_string:
    if i == 'b':
        b_string += i
    else:
        new_string += i
a_string = new_string

print(a_string)
```

Only the 1st letter of the shorter word should be capitalized.
Note: If the words are equal lengths, pick the 1st word as the longer one.

```python
phrase = str(input('Give me two words:'))

split_phrase = phrase.split(' ')
first = split_phrase[0]
second = split_phrase[1]
longer = len(first) > len(second)

for i in split_phrase:
    if longer is True:
        longer_word = first + '' 
        shorter_word = second + ''
    elif len(first) == len(second):
        longer_word = first + '' 
        shorter_word = second + ''
    else:
        shorter_word = first + ''
        longer_word = second + ''

print(longer_word.upper())
print(shorter_word.title())
```

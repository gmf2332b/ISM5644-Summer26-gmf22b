```python
import pandas as pd

# Converts the names into a standardized name so that students can match across both
# semesters regardless of the middle names

def convert_name(full_name):
    indiv_names = full_name.strip().lower().split()
    if len(indiv_names) >= 2:
        return indiv_names[0] + ' ' + indiv_names[-1]
    return full_name.strip().lower()

# load roster into memory so no need to hardcode it each time (turned into a dictionary)

def load_roster():
    semester1_file = open('semester1.txt','r')
    roster1 = {}
    for student in semester1_file:
        if student.strip():
            indiv_names = student.strip().split(';')
            full_name = indiv_names[0]
            converted = convert_name(full_name)
            score = [int(num.strip()) for num in indiv_names[1:] if num.strip().isdigit()]
            roster1[converted] = score
    semester1_file.close()

    semester2_file = open('semester2.txt','r')
    roster2 = {}
    for student in semester2_file:
        if student.strip():
            indiv_names = student.strip().split(';')
            full_name = indiv_names[0]
            converted = convert_name(full_name)
            score = [int(n.strip()) for n in indiv_names[1:] if n.strip().isdigit()]
            roster2[converted] = score
    semester2_file.close()
    return roster1,roster2

# finding students who were in semester 1 but not in semester 2

def find_dropped(roster1,roster2):
    dropped_students = []
    for name in roster1:
        if name not in roster2:
            dropped_students.append(name)
    return dropped_students

# finding the summary statistics for a student

def summary_statistics(semesterid,roster1,roster2):
    if semesterid == 1:
        chosen_roster = roster1
    elif semesterid == 2:
        chosen_roster = roster2
    else:
        print("Semester ID doesn't exist")
        return []
    df = pd.DataFrame.from_dict(chosen_roster,orient='index',columns=['Exam 1','Exam 2'])
    stats = pd.DataFrame({'Count':df.count(),'Average': df.mean(),'Standard Deviation': df.std()})
    return stats
    
# finding the exam scores for students for both semesters


def student_performance(studentName,roster1,roster2):
    semester1_score = roster1.get(convert_name(studentName),[])
    semester2_score = roster2.get(convert_name(studentName),[])
    return {'Semester 1': semester1_score,'Semester 2': semester2_score}

# A cutoff score is asked, then returns students who scored lower than the cutoff score

def cutoff_report(cutoffScore,semesterid,examid,roster1,roster2):
    if semesterid == 1:
        chosen_roster = roster1
    elif semesterid == 2:
        chosen_roster = roster2
    else:
        print("Semester ID doesn't exist")
        return []
        
    exam_index = examid - 1
    below_cutoff = []

    for student,score in chosen_roster.items():
        if len(score) > exam_index:
            exam_score = score[exam_index]
            if exam_score < cutoffScore:
                below_cutoff.append(student.title())
    return below_cutoff

# To tie it all together, the main function

def main():
    roster1,roster2 = load_roster()

    run = True
    while run:
        print('\nRoster Management:')
        print('1. Find students that dropped')
        print('2. See summary statistics for students')
        print('3. See student performance')
        print('4. Cutoff report')
        print('5. Never mind')

        select = input('\nSelect an option:')

        if select == '1':
            dropped = find_dropped(roster1,roster2)
            print('\nDropped Students:')
            for student in dropped:
                print(student.title())
                
        elif select == '2':
            semester_int = int(input('What is the semester ID? (1 or 2):'))
            if semester_int not in range(1,3):
                print("Semester ID doesn't exist")
            else:
                stats = summary_statistics(semester_int,roster1,roster2)
                print('Stats for Semester',semester_int)
                print(stats)
                    
        elif select == '3':
            name = input("What is the student's name?:")
            performance = student_performance(name,roster1,roster2)
            if performance:
                s1_perf = performance['Semester 1']
                s2_perf = performance['Semester 2']
                print('Scores for',name.title())
                print('Semester 1:',performance['Semester 1'])
                print('Semester 2:',performance['Semester 2'])
                if s1_perf and not s2_perf:
                    print('Student was dropped from the roster.')
                
                
        elif select == '4':
            cutoff = int(input('What is the cutoff score?:'))
            semester_int = int(input('What is the semester ID? (1 or 2):'))
            if semester_int not in range(1,3):
                print("Semester ID doesn't exist")
            else:
                exam_int = int(input('What is the exam ID? (1 or 2):'))
                result = cutoff_report(cutoff,semester_int,exam_int,roster1,roster2)
                print('\nStudents below cutoff:',cutoff)
                if result:
                    for student in result:
                        print(student)
                else:
                    print('None')

        elif select == '5':
            print('Exiting the program, see ya!')
            break

        else:
            print('\nError, select an option 1-5')
            continue

        run_again = input('\nRun another option, yes or no?').strip().lower()
        #if run_again == 'no':
            #print('See ya!')
            #break

        try:
            if run_again not in ['yes','y','no','n']:
                print('Not valid.')
            if run_again in ['no','n']:
                print('See ya!')
                break
        except ValueError:
            print('Incorrect value, try again.')
```

```python
main()
```

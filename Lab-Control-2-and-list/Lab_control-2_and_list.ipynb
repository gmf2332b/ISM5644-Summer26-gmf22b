#### 1. To find the average of a sequence in a list, we can sum the items and divide the sum to the number of items in the list. Find the average value in the following list: [9,41,12,3,74,11] and print it out.
#### Note: You should only use “+” and “/” arithmetic operators and a *for* loop – but no other function or method from any packages.

```python
alist = [9,41,12,3,74,11]
total = 0
counter = 0

for i in alist:
    total += i
    counter += 1
average = total / counter
print(average)
```

#### 5. Write a program to calculate and print the sum of all prime numbers from 1 to n (n inclusive) where n is a number input by the user.
#### Hint: Consider using a nested loop. The outer loop can be useful for iterating over the numbers between 1 and n and the inner loop can be used to check if the current number in the iteration is prime or not.

```python
n = int(input('number:'))
total = 0

for i in range (2,n+1):
    prime = True
    for j in range (2,i):
        if i % j == 0:
            prime = False
            break
    if prime:
        total += 1

print('sum of all prime numbers up to', n, 'is:', total)   
```

#### 10. Use a *list comprehension* expression to create a list of all the factors of a number that is entered by the user. Factors are those that divide the number as a whole – i.e., without leaving a remainder. For example, if you input 12, then the result should be [1,2,3,4,6,12]

#### 11. Given two lists such as:
* a_list = [1,2,4]
* b_list = [5,23,6,1,8,1,2,35,123,3]
#### Write a program to check if each element in a_list occurs in b_list or not. If they are all in the b_list, then print out 'All elements are in the b_list.' Otherwise, print out, 'Not all elements are included in the b_list.' . Test your program with <u>different inputs</u>.

```python
a_list = [1,2,4]
b_list = [5,23,6,1,8,1,2,35,123,3]
included = True

for i in a_list:
    found_i = False
    for j in b_list:
        if i == j:
            found_i = True
            break
    if found_i == False:
        included = False
        break
if included:
    print('All elements are in the b_list')
else: 
    print('Not all elements are included in the b_list')
```
